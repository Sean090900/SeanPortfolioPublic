```python
import numpy as np
from PIL import Image
from modules.extraction.preprocessing import Preprocessing
from modules.extraction.embedding import Embedding
from modules.retrieval.index.bruteforce import FaissBruteForce
from modules.retrieval.index.hnsw import FaissHNSW
from modules.retrieval.index.lsh import FaissLSH
from modules.retrieval.search import FaissSearch

print('DONE')
```

```python
## List of designed parameters: 
MODEL = 'casia-webface'
INDEX = 'bruteforce'
SIMILARITY_MEASURE = 'cosine'

# Object Initialization
preprocessor = Preprocessing()
model = Embedding(MODEL)
index = FaissBruteForce(dim=512, metric=SIMILARITY_MEASURE)
# index = FaissHNSW(dim=512, metric=SIMILARITY_MEASURE)
# index = FaissLSH(dim=512)
```

```python
# For entire dataset (probe dir), preprocess and retrieve embeddings
dataset_path = 'storage/multi_image_gallery'
embeddings = []
metadata = []
for image_set in os.listdir(dataset_path):
    for image_name in os.listdir(dataset_path + '/' + image_set):
        print(image_name)
        path = dataset_path + '/' + image_set + '/' + image_name
        image = Image.open(path)
        processed_tensor = preprocessor.process(image)
        embeddings.append(model.encode(processed_tensor))
        metadata.append(image_name)
```

```python
# Add embeddings and save index
index.add_embeddings(embeddings, metadata)
index.save(f'storage/embeddings/{MODEL}/faiss_{INDEX}_{MODEL}_cosine_index.pkl')
```
