```python
# Imports
from modules.inference.model import Detector
from modules.utils.loss import Loss
from modules.rectification.hard_negative_mining import HardNegativeMiner

# Static Parameters
WEIGHTS_PATH = 'storage/yolo_model_1/yolov4-tiny-logistics_size_416_1.weights'
CONFIG_PATH = "storage/yolo_model_1/yolov4-tiny-logistics_size_416_1.cfg"
CLASS_NAMES_PATH = "storage/yolo_model_1/logistics.names"
SCORE_THRESHOLD = 0.5
IOU_THRESHOLD = 0.2
NUM_CLASSES = 20

# Variable Parameters
LAMBDA_COORD = 0.5
LAMBDA_OBJ = 0.5
LAMBDA_NOOBJ = 0.5
LAMBDA_CLS = 100

print('DONE')
```

```python
# Inititialize objects
dataset_dir = 'storage/sampled_dataset_5000'
model = Detector(WEIGHTS_PATH, CONFIG_PATH, CLASS_NAMES_PATH, SCORE_THRESHOLD)
loss = Loss(IOU_THRESHOLD, LAMBDA_COORD, LAMBDA_OBJ, LAMBDA_NOOBJ, LAMBDA_CLS, NUM_CLASSES)
hnm = HardNegativeMiner(model, loss, dataset_dir)

df = hnm.sample_hard_negatives(5, 'total_loss')
```

```python
# Were using YOLO Loss as a metric for sampling hard negative images
# changing lambda values should change images sampled, based on what those lambda values represent
# The lambda values are weights applyed to different loss values, making those values more important when it comes to sampling hard negatives 
# 
# Raising LAMBDA_COORD: This weights the "coordinate loss", and samples images where the model was worst at creating bounding boxes around objects
# Raising LAMBDA_OBJ: This weights "objectness loss", and samples images where the model was most unsuccessful in predicting if objects exist 
# Raising LAMBDA_NOOBJ: This weights "no-object loss", and samples images where the model was unsuccessfully predicting that objects existed when they did not
# Raising LAMBDA_CLS: This wieghts "classification loss", and samples images where the model was worst at predicting the type of objects present
# We're sampling images where the model is worst at classifying object. This will help the model prioritze classification of objects



```
